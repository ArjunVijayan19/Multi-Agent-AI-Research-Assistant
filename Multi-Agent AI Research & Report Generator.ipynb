```python
from crewai import Agent,Task,Crew,LLM

local_llm=LLM(
    model="ollama/llama3.2:1b",
    base_url="http://localhost:11434"
)
```

```python
researcher=Agent(
    role='Researcher',
    goal="Identify 3 major trends in {topic}",
    backstory="You are a tech-savy researcher who find most relevant and up-to-date information",
    llm=local_llm,
    verbose=True
)
```

```python
writer=Agent(
    role='Writer',
    goal='Summarize the research findings into a short report',
    backstory='You are a professional technical writer known for being concise and clear',
    llm=local_llm,
    verbose=True
)
```

```python
research_task=Task(
    description='Search and Analyze about  news about {topic}',
    expected_output='A bulleted list of top 3 trends found.',
    agent=researcher
)
```

```python
write_task=Task(
    description="Write a two paragraph summary based on researcher's finding",
    expected_output='A clean formatted summary report',
    agent=writer
)
```

```python
crew=Crew(
    agents=[researcher,writer],
    tasks=[research_task,write_task],
    memory=False,
    verbose=True
)
```

```python
result=crew.kickoff(inputs={'topic':'Agentic AI'})
print("\n\n##############FINAL OUTPUT#############")
print(result)
```
